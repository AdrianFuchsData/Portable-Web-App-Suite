# Eine laufende Zelle anhalten

Führen Sie die nächste Zelle aus und drücken Sie dann **⏹** (oder `I`, `I` im Befehlsmodus). Vom Relay bereitgestellt, endet die Zelle mit einem `KeyboardInterrupt`, und `counter` behält seinen Wert; als Datei geöffnet, startet das Anhalten Python neu.

```python
import time

counter = 0
while True:
    counter += 1
    time.sleep(0.1)
```
