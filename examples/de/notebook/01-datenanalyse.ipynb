# Datenanalyse

Läuft **offline** – in der Suite (als Datei geöffnet) und in jeder Jupyter-Installation mit pandas, matplotlib, seaborn, SciPy und scikit-learn.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
df = pd.DataFrame({
    'region': rng.choice(['Nord', 'Süd', 'West'], 200),
    'units': rng.integers(1, 50, 200),
})
df['price'] = rng.normal(10, 2, 200).round(2)
df['revenue'] = df.units * df.price
df.head()
```

```python
df.groupby('region')[['units', 'revenue']].agg(['sum', 'mean']).round(1)
```

## Diagramme

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
sns.histplot(df.revenue, kde=True, ax=axes[0])
sns.boxplot(data=df, x='region', y='revenue', ax=axes[1])
plt.tight_layout()
```

## Statistik und ein Modell

```python
from scipy import stats

north, south = (df[df.region == r].revenue for r in df.region.unique()[:2])
t, p = stats.ttest_ind(north, south)
print(f't = {t:.2f}, p = {p:.3f}')
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

X, y = df[['units', 'price']], df.revenue
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
model = LinearRegression().fit(X_train, y_train)
print('R² =', round(model.score(X_test, y_test), 3))
```
