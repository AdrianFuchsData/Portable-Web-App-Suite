# Dateien in einem Archiv

Öffnen Sie zuerst **example-data.zip** (aus examples.zip) im Archivmanager. In der Suite wird das Archiv unter `/mnt` eingebunden; in einer gewöhnlichen Jupyter-Installation wird es neben dem Notebook entpackt. Nach jeder Zelle gehen Änderungen zurück ins Archiv (speichern Sie es im Archivmanager).

```python
import os

try:
    import suite                                  # in the suite
    folder = await suite.archives.mount('example-data.zip')
except ModuleNotFoundError:                       # regular Jupyter
    import zipfile
    zipfile.ZipFile('example-data.zip').extractall('example-data')
    folder = 'example-data'

sorted(os.listdir(folder))
```

```python
import pandas as pd

df = pd.read_csv(os.path.join(folder, 'measurements.csv'))
df.describe()
```

```python
summary = df.groupby('station').temperature.mean().round(1)
summary.to_csv(os.path.join(folder, 'summary.csv'))   # wird nach dieser Zelle ins Archiv geschrieben
summary
```
