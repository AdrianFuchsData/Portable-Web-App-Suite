# Die Shell

Der Reiter **Terminal** des Notebooks hat eine bash-ähnliche Shell auf dem Python-Dateisystem (`help` listet die Befehle; `less` und `nano` sind interaktiv). In Zellen macht `suite.shell.run` dasselbe; in einer gewöhnlichen Jupyter-Installation verwenden dieselben Zellen die Shell des Systems.

```python
try:
    import suite.shell
    async def sh(command):
        return await suite.shell.run(command)
except ModuleNotFoundError:      # regular Jupyter
    import subprocess
    async def sh(command):
        return subprocess.run(command, shell=True).returncode
```

```python
await sh("mkdir -p demo && cd demo && printf 'anna,wien,31\nbernd,graz,45\nclara,wien,28\n' > people.csv && cd ..")
await sh('cat demo/people.csv | sort -t, -k3 -n | cut -d, -f1')
```

```python
await sh('grep -c wien demo/people.csv; tar -czf demo.tgz demo && tar -tzf demo.tgz')
```

Die Skripte **bericht.sh** und **sicherung.sh** der Beispiele laufen mit `sh bericht.sh` im Terminal (als Beispiel geöffnet, werden sie in das Python-Dateisystem gelegt). Skripte in einem eingebundenen Archiv laufen aus `/mnt/<Archiv>/`.
