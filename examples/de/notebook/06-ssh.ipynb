# SSH und SFTP

In der Suite (vom Relay bereitgestellt) verwendet `suite.ssh.Connection` die SSH-Engine der Suite; in einer gewöhnlichen Jupyter-Installation läuft derselbe Code mit **Fabric**. test.rebex.net bietet nur SFTP (keine Befehle); für `run` verwenden Sie Ihren eigenen Server.

```python
try:
    from suite.ssh import Connection      # in the suite
except ModuleNotFoundError:
    from fabric import Connection         # pip install fabric
```

```python
with Connection('test.rebex.net', user='demo', connect_kwargs={'password': 'password'}) as c:
    sftp = c.sftp()
    print(sftp.listdir('.'))
    c.get('readme.txt', 'readme.txt')
print(open('readme.txt').read()[:200])
```
