# E-Mail und FTP mit der Standardbibliothek

Sockets brauchen das **vom Relay bereitgestellte** Notebook (`relay.py --serve`, die Kopfzeile zeigt 🔓). Der Code ist gewöhnliches Python – dieselben Zellen laufen in einer gewöhnlichen Jupyter-Installation. Der FTP-Server ist das öffentliche, schreibgeschützte Demo von Rebex (test.rebex.net); für E-Mail tragen Sie Ihren eigenen Server ein.

```python
from ftplib import FTP

with FTP('test.rebex.net', timeout=20) as ftp:
    ftp.login('demo', 'password')
    print(ftp.nlst())
    lines = []
    ftp.retrlines('RETR readme.txt', lines.append)
print('\n'.join(lines[:5]))
```

```python
import smtplib
from email.message import EmailMessage

SERVER, PORT, USER = 'mail.example.org', 587, 'anna@example.org'   # Ihr Server
PASSWORD = ''   # zum Senden ausfüllen

msg = EmailMessage()
msg['From'], msg['To'], msg['Subject'] = USER, USER, 'Aus einem Notebook gesendet'
msg.set_content('Hallo aus Python!')

if PASSWORD:
    with smtplib.SMTP(SERVER, PORT, timeout=30) as s:
        s.starttls()
        s.login(USER, PASSWORD)
        s.send_message(msg)
    print('sent')
else:
    print(msg)
```
