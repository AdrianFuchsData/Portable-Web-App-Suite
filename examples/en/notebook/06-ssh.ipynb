# SSH and SFTP

In the suite (served by the relay) `suite.ssh.Connection` uses the suite's SSH engine; in a regular Jupyter installation the same code runs with **Fabric**. test.rebex.net offers SFTP only (no commands); for `run`, use your own server.

```python
try:
    from suite.ssh import Connection      # in the suite
except ModuleNotFoundError:
    from fabric import Connection         # pip install fabric
```

```python
with Connection('test.rebex.net', user='demo', connect_kwargs={'password': 'password'}) as c:
    sftp = c.sftp()
    print(sftp.listdir('.'))
    c.get('readme.txt', 'readme.txt')
print(open('readme.txt').read()[:200])
```
