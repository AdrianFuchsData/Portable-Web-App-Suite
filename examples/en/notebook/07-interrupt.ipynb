# Stopping a running cell

Run the next cell, then press **⏹** (or `I`, `I` in command mode). Served by the relay, the cell ends with a `KeyboardInterrupt` and `counter` keeps its value; opened as a file, stopping restarts Python.

```python
import time

counter = 0
while True:
    counter += 1
    time.sleep(0.1)
```
