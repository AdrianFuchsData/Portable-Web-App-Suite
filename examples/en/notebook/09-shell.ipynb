# The shell

The notebook's **Terminal** tab has a bash-like shell on the Python file system (`help` lists the commands; `less` and `nano` are interactive). In cells, `suite.shell.run` does the same; in a regular Jupyter installation the same cells use the system's shell.

```python
try:
    import suite.shell
    async def sh(command):
        return await suite.shell.run(command)
except ModuleNotFoundError:      # regular Jupyter
    import subprocess
    async def sh(command):
        return subprocess.run(command, shell=True).returncode
```

```python
await sh("mkdir -p demo && cd demo && printf 'anna,wien,31\nbernd,graz,45\nclara,wien,28\n' > people.csv && cd ..")
await sh('cat demo/people.csv | sort -t, -k3 -n | cut -d, -f1')
```

```python
await sh('grep -c wien demo/people.csv; tar -czf demo.tgz demo && tar -tzf demo.tgz')
```

The scripts **report.sh** and **backup.sh** of the examples run with `sh report.sh` in the Terminal (opened as an example, they are put into the Python file system). Scripts in a mounted archive run from `/mnt/<archive>/`.
