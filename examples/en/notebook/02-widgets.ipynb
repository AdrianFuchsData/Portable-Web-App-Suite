# Interactive widgets

Move the sliders – the plot follows.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

@interact(frequency=(1, 10), amplitude=(0.1, 2.0))
def wave(frequency=3, amplitude=1.0):
    x = np.linspace(0, 2 * np.pi, 400)
    plt.plot(x, amplitude * np.sin(frequency * x))
    plt.ylim(-2, 2)
    plt.show()
```
