In the suite, `requests` goes through the **relay** (set it in the Files panel: Network). This works also when the notebook is opened as a file. In Jupyter, requests connects directly.

```python
import requests

r = requests.get('https://httpbin.org/get', params={'lang': 'en'}, timeout=20)
r.raise_for_status()
r.json()['args']
```

```python
r = requests.post('https://httpbin.org/post', json={'name': 'Example'}, timeout=20)
r.json()['json']
```
