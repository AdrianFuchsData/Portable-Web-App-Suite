# Mail and FTP with the standard library

Sockets need the notebook **served by the relay** (`relay.py --serve`, the header shows 🔓). The code is plain Python – the same cells run in a regular Jupyter installation. The FTP server is the public read-only demo of Rebex (test.rebex.net); for mail, enter your own server.

```python
from ftplib import FTP

with FTP('test.rebex.net', timeout=20) as ftp:
    ftp.login('demo', 'password')
    print(ftp.nlst())
    lines = []
    ftp.retrlines('RETR readme.txt', lines.append)
print('\n'.join(lines[:5]))
```

```python
import smtplib
from email.message import EmailMessage

SERVER, PORT, USER = 'mail.example.org', 587, 'anna@example.org'   # your server
PASSWORD = ''   # fill in to send

msg = EmailMessage()
msg['From'], msg['To'], msg['Subject'] = USER, USER, 'Sent from a notebook'
msg.set_content('Hello from Python!')

if PASSWORD:
    with smtplib.SMTP(SERVER, PORT, timeout=30) as s:
        s.starttls()
        s.login(USER, PASSWORD)
        s.send_message(msg)
    print('sent')
else:
    print(msg)
```
