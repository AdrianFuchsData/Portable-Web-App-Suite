# Files in an archive

Open **example-data.zip** (from examples.zip) in the archive manager first. In the suite the archive is mounted below `/mnt`; in a regular Jupyter installation it is unpacked next to the notebook. After every cell, changes go back into the archive (save it in the archive manager).

```python
import os

try:
    import suite                                  # in the suite
    folder = await suite.archives.mount('example-data.zip')
except ModuleNotFoundError:                       # regular Jupyter
    import zipfile
    zipfile.ZipFile('example-data.zip').extractall('example-data')
    folder = 'example-data'

sorted(os.listdir(folder))
```

```python
import pandas as pd

df = pd.read_csv(os.path.join(folder, 'measurements.csv'))
df.describe()
```

```python
summary = df.groupby('station').temperature.mean().round(1)
summary.to_csv(os.path.join(folder, 'summary.csv'))   # written into the archive after this cell
summary
```
